### Removing the zero-varience columns
- These column has zero dominance

```python
dftrain.drop("ID", axis=1, inplace=True)
dftest.drop("ID", axis=1, inplace=True)
```

```python
# List out the columns with only one unique value and remove them from both train and test datasets
removeingList = []
for val in dftrain.columns:
    if len(dftrain[val].value_counts()) < 2:
        # print(val, len(dftrain[val].value_counts()))
        dftrain.drop(val, axis=1, inplace=True)
        dftest.drop(val, axis=1, inplace=True)
```

### Dummy encoding all columns

```python
dftrain = pd.get_dummies(dftrain, dtype=int, columns=["X0", "X2", "X3", "X4", "X5", "X6", "X1", "X8"], drop_first=True)
dftest= pd.get_dummies(dftest, dtype=int, columns=["X0", "X2", "X3", "X4", "X5", "X6", "X1", "X8"], drop_first=True)
```

### Dimensionality Reduction

```python
dftest = dftest.reindex(columns=dftrain.columns, fill_value=0) # * reindex dftest to match dftrain columns
pca = PCA(n_components=8,random_state=42) # n_c
dftrain_pca = pca.fit_transform(dftrain)
dftest_pca = pca.transform(dftest)
```
